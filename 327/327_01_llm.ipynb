### NLP (자연어 처리)

```python
# 사전 설치 : pip install konlpy
# jdk11 사전설치필요(환경변수 설정 포함)
from konlpy.tag import Okt    # 한국어 자연어처리 라이브러리 (형태소 분석)
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding
from tensorflow.keras.models import Sequential

# 1. 텍스트 데이터 (입력 문장)
sentences = [
    "자연어 처리는 재미있는 분야입니다.",
    "딥러닝은 많은 데이터를 필요로 합니다.",
    "한국어 NLP는 정말 재미있어요!"
]

# 2. 토크나이징 (Tokenizing)
okt = Okt()
tokenized_sentences = [okt.morphs(sentence) for sentence in sentences]
print("토크나이징 결과:", tokenized_sentences)

# 3. 인코딩 (Encoding): 단어를 숫자로 변환
tokenizer = Tokenizer()
tokenizer.fit_on_texts(tokenized_sentences)
encoded_sentences = tokenizer.texts_to_sequences(tokenized_sentences)
print("인코딩 결과:", encoded_sentences)

# 4. 패딩 (Padding): 길이를 맞추기 위해 0으로 채우기
max_len = 10  # 최대 길이 설정
padded_sentences = pad_sequences(encoded_sentences, maxlen=max_len, padding='post')
print("패딩 결과:", padded_sentences)

# 5. 임베딩 (Embedding)
vocab_size = len(tokenizer.word_index) + 1  # 단어 사전 크기
embedding_dim = 8  # 임베딩 차원 크기

# 간단한 임베딩 모델 생성
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_len))
model.compile('rmsprop', 'mse')

# 패딩된 문장을 임베딩 층에 통과
embeddings = model.predict(padded_sentences)
print("임베딩 결과 (첫 번째 문장):\n", embeddings[0])
```

```python
# 감정분석(zero-shot classification)
### transformers 라이브러리 사전 설치 : pip install transformers
### tf-keras 라이브러리 사전 설치치 : pip install tf-keras
from transformers import pipeline
"""
1. feature-extraction: 특징 추출 (텍스트에 대한 벡터 표현 추출)
2. fill-mask : 마스크 채우기
3. ner : 개체명 인식 (named entity recognition)
4. question-answering : 질의응답
5. sentiment-analysis : 감정 분석
6. summarization : 요약
7. text-generation : 텍스트 생성
"""
classifier = pipeline("sentiment-analysis") # 5. 감정 분석
classifier("오늘 기분이 좋아요")
```

```python
classifier(
    ["I've been waiting for a HuggingFace course my whole life.", "I hate this so much!"]
)
```

```python
classifier(
    ["이 문제는 너무 어려워.", "너무 고마워서 어떡하지?"] # 
)
```

```python
# 텍스트 생성(text generation)
generator = pipeline("text-generation") # 7. text-generation : 텍스트 생성
generator("In this course, we will teach you how to")
```

```python
generator("today, i'm going to")
```

```python
# question-answering
question_answerer = pipeline("question-answering") # 4. 질의응답
question_answerer(
    question="Where do I work?",
    context="My name is Sylvain and I work at Hugging Face in Brooklyn",
)
```

```python
question_answerer(
    question="Where is my house?",
    context="My name is Gildong and I work at Samsung in Seoul",
)
```
