## 데이터프레임 로드

```python
import pandas as pd

travel = "dataset/travel.csv"
traveller = "dataset/traveller_master.csv"
loc = "dataset/visit_area_info.csv"

travelDF = pd.read_csv(travel, encoding="utf-8")
travellerDF = pd.read_csv(traveller, encoding="utf-8")
locDF = pd.read_csv(loc, encoding="utf-8")

travelDF.head()
```

```python
travelDF_f = travelDF[["TRAVEL_ID","TRAVELER_ID"]]
travelDF_f.head()
```

```python
travellerDF.head()
```

```python
travellerDF_f = travellerDF[["TRAVELER_ID","GENDER","AGE_GRP"]]
travellerDF_f.head()
```

```python
locDF.head()
```

```python
locDF_f = locDF[["TRAVEL_ID","VISIT_AREA_NM","X_COORD","Y_COORD","DGSTFN"]]
locDF_f.head()
```

```python
travelDF_f1 = pd.merge(travelDF_f,travellerDF_f, on="TRAVELER_ID", how="left")
travelDF_f1.head()
```

```python
travelDF_f2 = pd.merge(locDF_f, travelDF_f1, on="TRAVEL_ID", how="left")
travelDF_f2.head()
```

```python
#print(travelDF_f2.isnull().sum()) # 결측치 확인
"""
TRAVEL_ID           0
VISIT_AREA_NM       0
X_COORD          6927
Y_COORD          6927
DGSTFN           5540
TRAVELER_ID         0
GENDER              0
AGE_GRP             0
dtype: int64
"""
travelDF_f2.dropna(inplace=True) # 결측치 제거
print(travelDF_f2.isnull().sum()) # 결측치 확인
```

```python
finalDF = travelDF_f2.drop(columns=["TRAVEL_ID","TRAVELER_ID"])
finalDF.head()
```

```python
print(int(finalDF[["DGSTFN"]].values[1]))
```

```python
for idx,dgs in enumerate(finalDF[["DGSTFN"]].values):
    dgs_int = int(dgs)
    if dgs_int==1:
        finalDF.at[idx, 'DGSTFN'] = "매우 불만족"
    elif dgs_int==2:
        finalDF.at[idx, 'DGSTFN']="불만족"
    elif dgs_int==3:
        finalDF.at[idx, 'DGSTFN']="보통"
    elif dgs_int==4:
        finalDF.at[idx, 'DGSTFN']="만족"
    elif dgs_int==5:
        finalDF.at[idx, 'DGSTFN']="매우 만족"

finalDF.head()
```

```python
finalDF.columns=["방문지 이름","X 좌표","Y 좌표","만족도","방문자 성별","방문자 나이대"]
finalDF.head()
```
