```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

# 크롬드라이버 설치 + 실행
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)
driver.get("https://place.map.kakao.com/16856059")

time.sleep(3)  # JS가 로드될 시간 확보

html = driver.page_source
soup = BeautifulSoup(html,'html.parser')
```

```python
details1 = soup.find_all('span', {'class': 'txt_detail'})
details2 = soup.find_all('span', {'class': 'tit_detail'})
details3 = soup.find_all('span', {'class': 'txt_detail2'})
```

```python
print(soup)
```

```python
for i,d in enumerate(details1):
    print(f'{i+1} 번째 정보')
    print(d.get_text())
    print('--------------------')
```

```python
for i,d in enumerate(details2):
    print(f'{i+1} 번째 정보')
    print(d.get_text())
    print('--------------------')
```
