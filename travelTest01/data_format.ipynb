## 데이터 전처리

```python
import pandas as pd
```

```python
## 데이터 셋 호출
travel_df = pd.read_csv('./dataset/tn_travel_F.csv')
loc_df = pd.read_csv('./dataset/tn_visit_area_info_F.csv')
```

```python
## 데이터 확인
travel_df.head()
```

```python
# 필요 데이터 추출
travel_df = travel_df[['TRAVEL_ID', 'TRAVEL_START_YMD', 'TRAVEL_END_YMD', 'TRAVEL_MISSION_CHECK']]
loc_df = loc_df[['TRAVEL_ID', 'POI_NM', 'X_COORD', 'Y_COORD', 'DGSTFN', 'SGG_CD']]
```

```python
travel_df.head()
```

```python
loc_df.head()
```

```python
travelDF_f1 = pd.merge(loc_df,travel_df, on="TRAVEL_ID", how="left")
travelDF_f1.head()
```

```python
travelDF_f1.dropna(inplace=True) # 결측치 제거
print(travelDF_f1.isnull().sum()) # 결측치 확인
```

```python
travelDF_f1.head()
```

```python
travelDF_f1['TRAVEL_END_YMD'] = pd.to_datetime(travelDF_f1['TRAVEL_END_YMD'])
travelDF_f1['TRAVEL_START_YMD'] = pd.to_datetime(travelDF_f1['TRAVEL_START_YMD'])

travelDF_f1['date_diff'] = (travelDF_f1['TRAVEL_END_YMD'] - travelDF_f1['TRAVEL_START_YMD']).dt.days
travelDF_f1.drop(columns=['TRAVEL_START_YMD','TRAVEL_END_YMD'], inplace=True)
travelDF_f1.head()
```

```python
# 위치정보 필요부분만 남기기
travelDF_f1['SGG_CD'] = travelDF_f1['SGG_CD'].astype(str).str[:5].astype(float).astype(int)
travelDF_f1.head()
```

```python
# 중복 데이터 제거
travelDF_f1 = travelDF_f1.drop_duplicates()
travelDF_f1.head()
```

```python
# ID 값 제거
travelDF_f1.drop(columns=['TRAVEL_ID'], inplace=True)
travelDF_f1.head()
```

```python
# 미션값 리스트로
travelDF_f1['TRAVEL_MISSION_CHECK'] = travelDF_f1['TRAVEL_MISSION_CHECK'].str.split(';')
travelDF_f1.head()
```

```python
travelDF_f1.columns = ['위치명','X 좌표','Y 좌표','만족도', '지역코드','여행 목적','여행 일수']
travelDF_f1 = travelDF_f1[['위치명','X 좌표','Y 좌표','지역코드','여행 목적','여행 일수','만족도']]
travelDF_f1.head()
```

```python
travelDF_f1.to_csv('dataset/train_travel_F.csv', index=False)
```
